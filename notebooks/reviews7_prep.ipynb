```python
import pandas as pd
import numpy as np
```

```python
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
```

```python
print("Shape:", reviews.shape)
```

```python
print("\nFirst 5 rows:")
display(reviews.head())
```

```python
print("\nData types:")
print(reviews.dtypes)
```

```python

print("\nMissing values:")
print(reviews.isnull().sum())
```

```python
print("\nDuplicate rows:", reviews.duplicated().sum())
```

```python
reviews["review_score"] = pd.to_numeric(
    reviews["review_score"],
    errors="coerce"
)
```

```python
print("\nInvalid review scores:")
print(
    (
        (reviews["review_score"] < 1) |
        (reviews["review_score"] > 5)
    ).sum()
)
```

```python
reviews.loc[
    (reviews["review_score"] < 1) |
    (reviews["review_score"] > 5),
    "review_score"
] = np.nan
```

```python
reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"],
    errors="coerce"
)

reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"],
    errors="coerce"
)
```

```python
reviews["review_comment_title"] = (
    reviews["review_comment_title"]
    .astype("string")
    .str.strip()
)

reviews["review_comment_message"] = (
    reviews["review_comment_message"]
    .astype("string")
    .str.strip()
)
```

```python
reviews["review_category"] = np.select(
    [
        reviews["review_score"] <= 2,
        reviews["review_score"] == 3,
        reviews["review_score"] >= 4
    ],
    [
        "Negative",
        "Neutral",
        "Positive"
    ],
    default="Unknown"
)
```

```python
reviews = reviews.drop_duplicates().reset_index(drop=True)
```

```python
print("\nFinal shape:", reviews.shape)
```

```python
print("\nReview categories:")
print(reviews["review_category"].value_counts())
```

```python
reviews.to_csv(
    "../data/processed/reviews_cleaned.csv",
    index=False
)

print("\nReviews preprocessing completed.")
```
