```python
import pandas as pd
import numpy as np
```

```python
geo = pd.read_csv(
    "../data/raw/olist_geolocation_dataset.csv")
```

```python
print("Shape:", geo.shape)
```

```python
print("\nFirst 5 rows:")
display(geo.head())
```

```python
print("\nData types:")
print(geo.dtypes)
```

```python
print("\nMissing values:")
print(geo.isnull().sum())
```

```python
print("\nDuplicate rows:", geo.duplicated().sum())
```

```python
geo["geolocation_zip_code_prefix"] = pd.to_numeric(
    geo["geolocation_zip_code_prefix"],
    errors="coerce"
)
```

```python
geo["geolocation_lat"] = pd.to_numeric(
    geo["geolocation_lat"],
    errors="coerce"
)

geo["geolocation_lng"] = pd.to_numeric(
    geo["geolocation_lng"],
    errors="coerce"
)
```

```python
geo["geolocation_city"] = (
    geo["geolocation_city"]
    .astype("string")
    .str.strip()
    .str.upper()
)

geo["geolocation_state"] = (
    geo["geolocation_state"]
    .astype("string")
    .str.strip()
    .str.upper()
)
```

```python
print(
    "\nInvalid latitude:",
    (
        (geo["geolocation_lat"] < -90) |
        (geo["geolocation_lat"] > 90)
    ).sum()
)

print(
    "Invalid longitude:",
    (
        (geo["geolocation_lng"] < -180) |
        (geo["geolocation_lng"] > 180)
    ).sum()
)
```

```python
geo.loc[
    (geo["geolocation_lat"] < -90) |
    (geo["geolocation_lat"] > 90),
    "geolocation_lat"
] = np.nan

geo.loc[
    (geo["geolocation_lng"] < -180) |
    (geo["geolocation_lng"] > 180),
    "geolocation_lng"
] = np.nan
```

```python
geo = geo.drop_duplicates().reset_index(drop=True)

print("\nFinal shape:", geo.shape)

print("\nFinal missing values:")
print(geo.isnull().sum())
```

```python
geo.to_csv(
    "../data/processed/geolocation_cleaned.csv",
    index=False
)

print("\nGeolocation preprocessing completed.")
```
