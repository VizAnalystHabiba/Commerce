```python
import pandas as pd
import numpy as np
```

```python
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
```

```python
print("Shape:", payments.shape)
```

```python
print("\nFirst 5 rows:")
display(payments.head())
```

```python
print("\nData types:")
print(payments.dtypes)
```

```python
print("\nMissing values:")
print(payments.isnull().sum())
```

```python
print("\nDuplicate rows:", payments.duplicated().sum())
```

```python
payments["payment_sequential"] = pd.to_numeric(
    payments["payment_sequential"],
    errors="coerce"
)

payments["payment_installments"] = pd.to_numeric(
    payments["payment_installments"],
    errors="coerce"
)

payments["payment_value"] = pd.to_numeric(
    payments["payment_value"],
    errors="coerce"
)
```

```python
payments["payment_type"] = (
    payments["payment_type"]
    .astype("string")
    .str.strip()
    .str.lower()
)
```

```python
print(
    "\nNegative payment values:",
    (payments["payment_value"] < 0).sum()
)
```

```python
payments.loc[
    payments["payment_value"] < 0,
    "payment_value"
] = np.nan

payments.loc[
    payments["payment_installments"] < 0,
    "payment_installments"
] = np.nan
```

```python
payments = payments.drop_duplicates().reset_index(drop=True)

print("\nFinal shape:", payments.shape)

print("\nPayment types:")
print(payments["payment_type"].value_counts())

print("\nFinal missing values:")
print(payments.isnull().sum())
```

```python
payments.to_csv(
    "../data/processed/payments_cleaned.csv",
    index=False
)

print("\nPayments preprocessing completed.")
```
