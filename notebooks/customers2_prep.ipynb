```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
```

```python
print("Shape:", customers.shape)
```

```python
print("\nFirst 5 rows:")
display(customers.head())
```

```python
print("\nData types:")
print(customers.dtypes)
```

```python
print("\nMissing values:")
print(customers.isnull().sum())
```

```python
print("\nDuplicate rows:", customers.duplicated().sum())
```

```python
print("\nDuplicate customer IDs:")
print(customers["customer_id"].duplicated().sum())
```

```python
print("\nCustomer states:")
print(customers["customer_state"].value_counts())
```

```python
# Clean text columns
text_columns = [
    "customer_city",
    "customer_state"
]

for column in text_columns:
    customers[column] = customers[column].astype("string").str.strip().str.upper()
```

```python
# Convert ZIP code
customers["customer_zip_code_prefix"] = pd.to_numeric(
    customers["customer_zip_code_prefix"],
    errors="coerce"
)

# Check invalid ZIP values
print("\nMissing ZIP codes:")
print(customers["customer_zip_code_prefix"].isnull().sum())
```

```python
# Remove duplicate customer IDs
customers = customers.drop_duplicates(
    subset="customer_id"
).reset_index(drop=True)

# Final check
print("\nFinal shape:", customers.shape)

print("\nFinal missing values:")
print(customers.isnull().sum())
```

```python
# Save
customers.to_csv(
    "../data/processed/customers_cleaned.csv",
    index=False
)

print("\nCustomers preprocessing completed.")
```
