```python
import pandas as pd
```

```python
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")
```

```python
print("Shape:", sellers.shape)
```

```python
print("\nFirst 5 rows:")
display(sellers.head())
```

```python
print("\nData types:")
print(sellers.dtypes)
```

```python
print("\nMissing values:")
print(sellers.isnull().sum())
```

```python
print("\nDuplicate rows:", sellers.duplicated().sum())
```

```python
print("\nDuplicate seller IDs:")
print(sellers["seller_id"].duplicated().sum())
```

```python
sellers["seller_city"] = (
    sellers["seller_city"]
    .astype("string")
    .str.strip()
    .str.upper()
)

sellers["seller_state"] = (
    sellers["seller_state"]
    .astype("string")
    .str.strip()
    .str.upper()
)
```

```python
sellers["seller_zip_code_prefix"] = pd.to_numeric(
    sellers["seller_zip_code_prefix"],
    errors="coerce"
)
```

```python
sellers = sellers.drop_duplicates(
    subset="seller_id"
).reset_index(drop=True)

print("\nFinal shape:", sellers.shape)

print("\nFinal missing values:")
print(sellers.isnull().sum())
```

```python
sellers.to_csv(
    "../data/processed/sellers_cleaned.csv",
    index=False
)

print("\nSellers preprocessing completed.")
```
