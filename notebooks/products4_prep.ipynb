```python
import pandas as pd
import numpy as np
```

```python
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
```

```python
print("Shape:", products.shape)
```

```python

print("\nFirst 5 rows:")
display(products.head())
```

```python
print("\nData types:")
print(products.dtypes)
```

```python
print("\nMissing values:")
print(products.isnull().sum())
```

```python
print("\nDuplicate rows:", products.duplicated().sum())
```

```python
print("\nDuplicate product IDs:")
print(products["product_id"].duplicated().sum())
```

```python
numeric_columns = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for column in numeric_columns:
    products[column] = pd.to_numeric(
        products[column],
        errors="coerce"
    )
```

```python
products["product_category_name"] = (
    products["product_category_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)
```

```python
for column in numeric_columns:
    print(
        column,
        "negative values:",
        (products[column] < 0).sum()
    )
```

```python
for column in numeric_columns:
    products.loc[products[column] < 0, column] = np.nan
```

```python
products = products.drop_duplicates(
    subset="product_id"
).reset_index(drop=True)

print("\nFinal shape:", products.shape)

print("\nFinal missing values:")
print(products.isnull().sum())
```

```python
products.to_csv(
    "../data/processed/products_cleaned.csv",
    index=False
)

print("\nProducts preprocessing completed.")
```
