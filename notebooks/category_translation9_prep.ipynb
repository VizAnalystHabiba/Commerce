```python
import pandas as pd
```

```python
categories = pd.read_csv("../data/raw/product_category_name_translation.csv")
```

```python
print("Shape:", categories.shape)
```

```python
print("\nFirst 5 rows:")
display(categories.head())
```

```python
print("\nData types:")
print(categories.dtypes)
```

```python
print("\nMissing values:")
print(categories.isnull().sum())
```

```python
print("\nDuplicate rows:", categories.duplicated().sum())
```

```python
categories["product_category_name"] = (
    categories["product_category_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)

categories["product_category_name_english"] = (
    categories["product_category_name_english"]
    .astype("string")
    .str.strip()
    .str.lower()
)
```

```python
categories = categories.drop_duplicates(
    subset="product_category_name"
).reset_index(drop=True)

print("\nFinal shape:", categories.shape)

print("\nFinal missing values:")
print(categories.isnull().sum())
```

```python
categories.to_csv(
    "../data/processed/category_translation_cleaned.csv",
    index=False
)

print("\nCategory translation preprocessing completed.")
```
