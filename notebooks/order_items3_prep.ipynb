```python
import pandas as pd
import numpy as np
```

```python
items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
```

```python
print("Shape:", items.shape)
```

```python
print("\nFirst 5 rows:")
display(items.head())
```

```python
print("\nData types:")
print(items.dtypes)
```

```python

print("\nMissing values:")
print(items.isnull().sum())
```

```python
print("\nDuplicate rows:", items.duplicated().sum())
```

```python
items["order_item_id"] = pd.to_numeric(
    items["order_item_id"],
    errors="coerce"
)

items["price"] = pd.to_numeric(
    items["price"],
    errors="coerce"
)

items["freight_value"] = pd.to_numeric(
    items["freight_value"],
    errors="coerce"
)
```

```python
items["shipping_limit_date"] = pd.to_datetime(
    items["shipping_limit_date"],
    errors="coerce"
)

print("\nNegative prices:", (items["price"] < 0).sum())
print("Negative freight:", (items["freight_value"] < 0).sum())
```

```python

items.loc[items["price"] < 0, "price"] = np.nan
items.loc[items["freight_value"] < 0, "freight_value"] = np.nan

# Total item value
items["total_item_value"] = (
    items["price"] + items["freight_value"]
)
```

```python

items = items.drop_duplicates().reset_index(drop=True)

print("\nFinal shape:", items.shape)

print("\nMissing values:")
print(items.isnull().sum())
```

```python
items.to_csv(
    "../data/processed/order_items_cleaned.csv",
    index=False
)

print("\nOrder items preprocessing completed.")
```
